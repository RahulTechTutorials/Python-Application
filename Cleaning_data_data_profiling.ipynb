```python
df.age = df.age.apply(lambda x : 1 if x < 0 or pd.isna(x)  else x )
```

```python
df.age.isnull().sum()
```

```python
df.city.isnull().sum()
```

```python
df.city = df.city.apply(lambda x : df.city.mode()[0] if pd.isna(x) else x )
```

```python
signup_date_imputed_val = df.query("signupdate != 'not_a_date'").signupdate.mode()[0]
```

```python
df.signupdate = df.signupdate.apply(lambda x : signup_date_imputed_val if pd.isna(x) or x == 'not_a_date' else x)
```

```python
df.signupdate.isna().sum()
```

```python
df.signupdate = pd.to_datetime(df.signupdate, dayfirst=True)
```

```python
df.lastpurchasedate.isnull().sum()
```

```python
df.lastpurchasedate.value_counts().sort_values(ascending=False).head()
```

```python
lastpurchasedate_imputed_val = df.query( "lastpurchasedate != 'missing'").lastpurchasedate.mode()[0]
lastpurchasedate_imputed_val
```

```python
df.lastpurchasedate = df.lastpurchasedate.apply(lambda x : lastpurchasedate_imputed_val if x == 'missing' or pd.isna(x) else x)
```
