```python
##Generate two distributions and calcualte the Covariance between them


import numpy as np
import pandas as pd
```

```python
arr1 = np.random.randint(low = 10, high = 30, size = 1000)
```

```python
arr2 = np.random.randint(low = 20, high = 40, size = 1000)
```

```python
np.cov(arr1,arr2)
##1.96211812
```

Formula for Covariance = Sum((x - xbar) * (y - ybar)) / (s1 * s2 )


```python
n1_mean = np.mean(arr1)
n2_mean = np.mean(arr2)
n1_std = np.std(arr1)
n2_std = np.std(arr2)
n = len(arr1)
```

```python
n1_mean, n2_mean, n1_std, n2_std
```

```python
(arr2 - n2_mean).shape
```

```python
np.round(np.sum((arr1 - n1_mean) * (arr2 - n2_mean) ) / (n-1),8)
```
